```python
#PDF extraction 
def extract_pdf_pages(pdf_path: str) -> List[Dict]:
    """
    Extract page-level raw text from one PDF using pypdf.

    Returns a list of dicts:
        { source_pdf: str, page_num: int (1-indexed), raw_text: str }
    """
    reader = PdfReader(pdf_path)
    pages  = []
    for page_num, page in enumerate(reader.pages, start=1):
        text = page.extract_text() or ""
        pages.append({
            "source_pdf": os.path.basename(pdf_path),
            "page_num":   int(page_num),
            "raw_text":   text,
        })
    return pages
```

```python
# Text cleaning 
def clean_text(text: str) -> str:
    """
    Apply cleaning rules that address the most
    common PDF extraction.

    Rules (in order)
    ----------------
    1.  Unicode NFKC normalisation — resolves ligatures, half-width chars, etc.
    2.  Remove URLs and e-mail addresses (not useful for retrieval).
    3.  Remove browser-style timestamps  e.g. "4/16/26, 9:03 AM".
    4.  Remove page counters            e.g. "- 3 / 7 -", "9/9".
    5.  Re-join hyphenated line-breaks  e.g. "dis-\ntrubuted" → "distributed".
    6.  Collapse all remaining newlines into a single space.
    7.  Normalise Unicode punctuation to ASCII equivalents.
    8.  Strip non-printable control characters.
    9.  Truncate at reference/bibliography headers (irrelevant for retrieval).
    10. Collapse runs of whitespace to a single space.
    """
    if not text:
        return ""

    # 1. Unicode normalisation
    text = unicodedata.normalize("NFKC", text)

    # 2. URLs and e-mails
    text = re.sub(r"https?://\S+|www\.\S+",              " ", text)
    text = re.sub(r"\b[\w\.-]+@[\w\.-]+\.\w+\b",   " ", text)

    # 3. Browser timestamps
    text = re.sub(
        r"\b\d{1,2}/\d{1,2}/\d{2,4},?\s*\d{1,2}:\d{2}\s*(?:AM|PM|am|pm)\b",
        " ", text,
    )

    # 4. Page counters
    text = re.sub(r"\b-?\s*\d+\s*/\s*\d+\s*-?\b", " ", text)

    # 5. Hyphenated line-breaks
    text = re.sub(r"(\w)-\s*\n\s*(\w)", r"\1\2", text)

    # 6. Remaining newlines → space
    text = re.sub(r"\s*\n\s*", " ", text)

    # 7. Unicode punctuation normalisation
    text = (
        text.replace("\u2013", "-").replace("\u2014", "-")
            .replace("\u2018", "'").replace("\u2019", "'")
            .replace("\u201c", '"').replace("\u201d", '"')
            .replace("\u00a0", " ")
    )

    # 8. Control characters
    text = re.sub(r"[\x00-\x08\x0b\x0c\x0e-\x1f\x7f]", " ", text)

    # 9. Truncate at bibliography / references section header
    for marker in [r"\bReferences\b", r"\bBibliography\b",
                   r"\bWorks Cited\b", r"\bFurther Reading\b"]:
        m = re.search(marker, text, flags=re.IGNORECASE)
        if m:
            text = text[: m.start()]
            break

    # 10. Whitespace collapse
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
# Empty-page filter 
def is_empty_page(text: str,
                  min_words: int         = 20,
                  min_alpha_ratio: float = 0.50) -> bool:
    """
    Return True when a page carries so little useful text that chunking it
    would produce noisy, low-quality chunks.

    Criteria (all must pass for the page to be kept):
    * At least `min_words` whitespace-delimited tokens.
    * At least `min_alpha_ratio` of non-space characters are alphabetic
      (filters pages that are mostly numbers, symbols, or formulas).
    * At least 5 distinct word types (filters repetitive boilerplate).
    """
    if not text or not text.strip():
        return True

    words = text.strip().split()
    if len(words) < min_words:
        return True

    alpha_chars = sum(c.isalpha() for c in text)
    total_chars = sum(not c.isspace() for c in text)
    if total_chars == 0 or alpha_chars / total_chars < min_alpha_ratio:
        return True

    if len(set(words)) < 5:
        return True

    return False
```

```python
# Chunking with metadata
def chunk_page_text(
    source_pdf: str,
    page_num:   int,
    text:       str,
    chunk_size: int = CHUNK_SIZE,
    overlap:    int = CHUNK_OVERLAP,
    min_chunk_words: int = 50,
) -> List[Dict]:
    """
    Split one cleaned page into overlapping, fixed-size word chunks.

    Design choice
    ----------------
    chunk_size = 300 words
        A 300-word chunk maps to roughly 400 sub-word tokens, which fits
        within SBERT's 512-token limit and within a typical LLM
        context window entry.  Going below 200 words risks splitting
        definitions; going above 500 words approaches the model limit.

    overlap = 60 words  (20 % of chunk_size)
        The overlap window reduces the probability that a definition, formula,
        or worked example is split right at a boundary.  A 10-30 % overlap is
        the range specified by the project; so, 20 % is a good default.

    Metadata stored per chunk:
        id            – UUID for globally unique identification
        source_pdf    – file name of the originating PDF
        page_num      – 1-indexed page number
        chunk_id      – composite key  <pdf>_p<page>_c<chunk>
        start_word    – index of the first word in the original page token list
        end_word      – index of the last word  (exclusive end)
        chunk_text    – the raw text of the chunk
        chunk_word_count – word count 
    """
    if chunk_size <= 0:
        raise ValueError("chunk_size must be positive.")
    if overlap < 0 or overlap >= chunk_size:
        raise ValueError("overlap must satisfy 0 <= overlap < chunk_size.")

    words       = text.split()
    step        = chunk_size - overlap
    chunks      = []
    chunk_index = 0

    for start in range(0, len(words), step):
        end         = min(start + chunk_size, len(words))
        chunk_words = words[start:end]

        if len(chunk_words) < min_chunk_words:
            # Skip tail fragments that are too short to be informative
            continue

        chunks.append({
            "id":              str(uuid.uuid4()),
            "source_pdf":      source_pdf,
            "page_num":        int(page_num),
            "chunk_id":        f"{source_pdf}_p{page_num:03d}_c{chunk_index:03d}",
            "start_word":      int(start),
            "end_word":        int(end),
            "chunk_text":      " ".join(chunk_words),
            "chunk_word_count": int(len(chunk_words)),
        })
        chunk_index += 1

    return chunks
```

```python
# Distributed ingestion via PySpark RDD
def list_pdf_paths(input_dir: str = INPUT_DIR) -> List[str]:
    """Recursively collect all .pdf / .PDF paths under input_dir."""
    return sorted(
        str(p) for p in Path(input_dir).rglob("*")
        if p.is_file() and p.suffix.lower() == ".pdf"
    )


def pdf_to_chunks(
    spark:       SparkSession,
    input_dir:   str = INPUT_DIR,
    parallelism: int = PARALLELISM,
    chunk_size:  int = CHUNK_SIZE,
    overlap:     int = CHUNK_OVERLAP,
):
    """
    Distributed ingestion pipeline.

    Architecture
    ------------
    1. Collect all PDF file paths o.
    2. Parallelize the path list across `parallelism` Spark partitions.
    3. Each executor independently:
       a. reads pages with pypdf
       b. cleans each page
       c. filters empty pages
       d. chunks the remaining pages
    4. flatMap flattens partition outputs into a single RDD of chunk dicts.

    chunk_size and overlap are broadcast to avoid repeated serialisation
    of the closure on every task dispatch.
    """
    pdf_paths = list_pdf_paths(input_dir)
    if not pdf_paths:
        raise FileNotFoundError(f"No PDFs found in {input_dir!r}")

    log.info("Found %d PDF(s) in %s", len(pdf_paths), input_dir)

    sc = spark.sparkContext
    cs = sc.broadcast(chunk_size)
    ov = sc.broadcast(overlap)

    def process_pdf(path: str) -> List[Dict]:
        out = []
        try:
            for page in extract_pdf_pages(path):
                text = clean_text(page["raw_text"])
                if is_empty_page(text):
                    continue
                out.extend(
                    chunk_page_text(
                        source_pdf = page["source_pdf"],
                        page_num   = page["page_num"],
                        text       = text,
                        chunk_size = cs.value,
                        overlap    = ov.value,
                    )
                )
        except Exception as exc:
            # Isolate failures to individual PDFs; the rest of the job continues.
            print(f"WARNING: could not process {path}: {exc}")
        return out

    rdd   = sc.parallelize(pdf_paths, numSlices=parallelism).flatMap(process_pdf)
    count = rdd.count()
    log.info("Total chunks produced: %d", count)
    return rdd
```

```python
# Run distributed ingestion and inspect the results
chunk_rdd = pdf_to_chunks(
    spark       = spark,
    input_dir   = INPUT_DIR,
    parallelism = PARALLELISM,
    chunk_size  = CHUNK_SIZE,
    overlap     = CHUNK_OVERLAP,
)

# Collect to driver for embedding (sklearn / gensim / SBERT run on the driver)
chunks = chunk_rdd.collect()

# Summary statistics
_sources = {}
for ch in chunks:
    _sources.setdefault(ch["source_pdf"], []).append(ch["chunk_word_count"])

print(f"{'PDF':<40} {'pages (chunks)':<16} {'avg words/chunk'}")
print("-" * 70)
for pdf, wcs in sorted(_sources.items()):
    print(f"{pdf:<40} {len(wcs):<16} {sum(wcs)/len(wcs):.0f}")
print("-" * 70)
print(f"{'TOTAL':<40} {len(chunks)}")
```

```python
#TF-IDF embedding 
def embed_tfidf(
    chunks:       List[Dict],
    max_features: int = TFIDF_MAX_FEATURES,
    svd_dim:      int = TFIDF_SVD_DIM,
    min_df:       int = 2,
    output_dir:   str = OUTPUT_DIR,
) -> List[Dict]:
    """
    Fit a TF-IDF + TruncatedSVD (LSA) pipeline on the full chunk corpus
    and attach a dense embedding to every chunk.


    """
    if not chunks:
        raise ValueError("No chunks provided to embed_tfidf().")

    texts = [c["chunk_text"] for c in chunks]

    tfidf = TfidfVectorizer(
        max_features = max_features,
        sublinear_tf = True,        # apply 1+log(tf) scaling
        ngram_range  = (1, 2),      # unigrams + bigrams
        min_df       = min_df,      # ignore very rare terms
        stop_words   = "english",
    )

    log.info("Fitting TF-IDF  max_features=%s  min_df=%s", max_features, min_df)
    X_tfidf    = tfidf.fit_transform(texts)
    vocab_size = X_tfidf.shape[1]

    if vocab_size < 2:
        raise ValueError(f"TF-IDF vocabulary too small ({vocab_size}). "
                         "Lower min_df or add more documents.")

    n_components = min(svd_dim, vocab_size - 1)
    svd = TruncatedSVD(n_components=n_components, random_state=42)

    log.info("Fitting TruncatedSVD  n_components=%d", n_components)
    vectors            = svd.fit_transform(X_tfidf).astype(np.float32)
    explained_variance = float(svd.explained_variance_ratio_.sum())

    pipeline = Pipeline([("tfidf", tfidf), ("svd", svd)])

    for chunk, vec in zip(chunks, vectors):
        chunk["embedding"]          = vec.tolist()
        chunk["embedding_strategy"] = "tfidf"
        chunk["embedding_dim"]      = int(n_components)

    os.makedirs(output_dir, exist_ok=True)
    pipeline_path = os.path.join(output_dir, "tfidf_pipeline.pkl")
    with open(pipeline_path, "wb") as fh:
        pickle.dump(pipeline, fh)

    log.info(
        "TF-IDF pipeline saved | vocab=%d | svd_dim=%d | explained_var=%.3f",
        vocab_size, n_components, explained_variance,
    )
    return chunks
```

```python
# Word2Vec embedding 
def embed_word2vec(
    chunks:     List[Dict],
    dim:        int = WORD2VEC_DIM,
    output_dir: str = OUTPUT_DIR,
) -> List[Dict]:
    """
    Train a Word2Vec (skip-gram) model on the chunk corpus, then represent
    each chunk as the mean of its in-vocabulary word vectors.

    Why Word2Vec?
    -------------
    Word2Vec captures local syntactic and semantic similarity better than
    TF-IDF (e.g. 'distributed' ≈ 'parallelised') while being much lighter
    than full sentence transformers.

    OOV handling: chunks where every token falls outside the trained
    vocabulary receive a zero vector (logged as a warning).

    
    """
    from gensim.models import Word2Vec as GensimW2V

    if not chunks:
        raise ValueError("No chunks provided to embed_word2vec().")

    sentences     = [c["chunk_text"].lower().split() for c in chunks]
    non_empty     = [s for s in sentences if s]
    if not non_empty:
        raise ValueError("No tokens found for Word2Vec training.")

    log.info("Training Word2Vec on %d sentences  vector_size=%d", len(non_empty), dim)
    model = GensimW2V(
        sentences   = non_empty,
        vector_size = dim,
        window      = 5,
        min_count   = 2,
        workers     = 4,
        epochs      = 10,
        seed        = 42,
    )

    zero_vec  = np.zeros(dim, dtype=np.float32)
    oov_count = 0

    for chunk, sent in zip(chunks, sentences):
        vecs = [model.wv[w] for w in sent if w in model.wv]
        if vecs:
            vec = np.mean(vecs, axis=0).astype(np.float32)
        else:
            vec = zero_vec.copy()
            oov_count += 1
        chunk["embedding"]          = vec.tolist()
        chunk["embedding_strategy"] = "word2vec"
        chunk["embedding_dim"]      = int(dim)

    if oov_count:
        log.warning(
            "%d/%d chunks contained only OOV tokens — zero vector assigned.",
            oov_count, len(chunks),
        )

    os.makedirs(output_dir, exist_ok=True)
    model_path = os.path.join(output_dir, "word2vec.model")
    model.save(model_path)
    log.info("Word2Vec model saved to %s", model_path)
    return chunks
```

```python
#SBERT embedding 
def embed_sbert(
    chunks:     List[Dict],
    model_name: str = SBERT_MODEL,
    batch_size: int = 64,
    output_dir: str = OUTPUT_DIR,
) -> List[Dict]:
    """
    Encode all chunks with a pre-trained SentenceTransformer model.

    Why SBERT?
    ----------
    Models from the sentence-transformers library (all-MiniLM-L6-v2 chosen
    here) produce contextual, dense embeddings that capture long-range
    semantic relationships — outperforming both TF-IDF and Word2Vec on
    semantic retrieval benchmarks. .

    Note on distribution
    --------------------
    SentenceTransformer.encode() is a batched GPU/CPU call that is already
    internally parallelised. We run it on the Spark driver after a
    .collect() — calling it from within an executor would cause OOM errors
    and serialisation issues with the tokeniser.

    .
    """
    from sentence_transformers import SentenceTransformer

    if not chunks:
        raise ValueError("No chunks provided to embed_sbert().")

    texts = [c["chunk_text"] for c in chunks]

    log.info("Loading SBERT model: %s", model_name)
    model = SentenceTransformer(model_name)

    log.info("Encoding %d chunks  batch_size=%d", len(texts), batch_size)
    t0      = time.time()
    vectors = model.encode(
        texts,
        batch_size        = batch_size,
        show_progress_bar = True,
        convert_to_numpy  = True,
    ).astype(np.float32)
    elapsed     = time.time() - t0
    emb_dim     = int(vectors.shape[1])

    log.info("SBERT done in %.1fs  embedding_dim=%d", elapsed, emb_dim)

    for chunk, vec in zip(chunks, vectors):
        chunk["embedding"]          = vec.tolist()
        chunk["embedding_strategy"] = "sbert"
        chunk["embedding_dim"]      = emb_dim

    os.makedirs(output_dir, exist_ok=True)
    config_path = os.path.join(output_dir, "sbert_config.json")
    with open(config_path, "w") as fh:
        json.dump({"model_name": model_name, "embedding_dim": emb_dim,
                   "batch_size": batch_size}, fh, indent=2)

    log.info("SBERT config saved to %s", config_path)
    return chunks
```

```python
# Parquet schema and storage 
# Schema matches the spec:
# [id, source_pdf, page_num, chunk_id, start_word, end_word,
#  chunk_text, embedding, embedding_strategy]
# plus chunk_word_count and embedding_dim.

CHUNK_SCHEMA = StructType([
    StructField("id",                 StringType(),           False),
    StructField("source_pdf",         StringType(),           False),
    StructField("page_num",           IntegerType(),          False),
    StructField("chunk_id",           StringType(),           False),
    StructField("start_word",         IntegerType(),          False),
    StructField("end_word",           IntegerType(),          False),
    StructField("chunk_text",         StringType(),           False),
    StructField("chunk_word_count",   IntegerType(),          False),
    StructField("embedding",          ArrayType(FloatType()), False),
    StructField("embedding_strategy", StringType(),           False),
    StructField("embedding_dim",      IntegerType(),          False),
])


def save_to_parquet(
    spark:      SparkSession,
    chunks:     List[Dict],
    output_dir: str = OUTPUT_DIR,
    strategy:   str = EMBEDDING_STRATEGY,
) -> None:
    """
    Persist embedded chunks as Parquet, partitioned by embedding strategy.

    The output path  <output_dir>/strategy=<name>  lets Task 2 load only
    the partitions it needs without reading unrelated strategies.

    """
    if not chunks:
        raise ValueError("No chunks to save.")

    os.makedirs(output_dir, exist_ok=True)
    df       = spark.createDataFrame(chunks, schema=CHUNK_SCHEMA)
    out_path = os.path.join(output_dir, f"strategy={strategy}")

    log.info("Writing %d rows → %s", df.count(), out_path)
    (
        df.repartition(PARALLELISM)
          .write
          .mode("overwrite")
          .parquet(out_path)
    )

    verify = spark.read.parquet(out_path)
    log.info(
        "Verification — rows=%d | schema=%s",
        verify.count(), verify.schema.simpleString(),
    )
    verify.show(5, truncate=90)
```

```python
#Full pipeline 
EMBEDDING_DISPATCH = {
    "tfidf":    embed_tfidf,
    "word2vec": embed_word2vec,
    "sbert":    embed_sbert,
}


def run_pipeline(
    parallelism:        int = PARALLELISM,
    chunk_size:         int = CHUNK_SIZE,
    chunk_overlap:      int = CHUNK_OVERLAP,
    embedding_strategy: str = EMBEDDING_STRATEGY,
    input_dir:          str = INPUT_DIR,
    output_dir:         str = OUTPUT_DIR,
    **embedding_kwargs,
) -> List[Dict]:
    """
    End-to-end Task 1 pipeline.

    Steps
    -----
    1. Start / reuse Spark.
    2. Distributed PDF ingestion: path parallelisation + flatMap (PySpark RDD).
    3. Collect chunks to driver (required for sklearn / gensim / SBERT fit).
    4. Generate embeddings on driver.
    5. Persist as Parquet (HDFS or local FS).

    Parameters
    ----------
    parallelism        : Spark default parallelism and number of RDD partitions.
    chunk_size         : Number of words per chunk (200-500).
    chunk_overlap      : Overlap in words between consecutive chunks (10-30 %).
    embedding_strategy : One of 'tfidf', 'word2vec', 'sbert'.
    input_dir          : Directory containing the source PDF files.
    output_dir         : Root directory for Parquet output and model artefacts.
    **embedding_kwargs : Forwarded to the chosen embedding function
                         (e.g. max_features for TF-IDF).

    Returns
    -------
    List of embedded chunk dicts (also written to Parquet).
    """
    if embedding_strategy not in EMBEDDING_DISPATCH:
        raise ValueError(
            f"Unknown embedding_strategy {embedding_strategy!r}. "
            f"Choose from {list(EMBEDDING_DISPATCH)}."
        )

    t_start = time.time()
    spark   = build_spark(parallelism)

    # ── Step 1: distributed PDF ingestion 
    log.info("=== Step 1/4: Distributed PDF ingestion ===")
    chunk_rdd = pdf_to_chunks(spark, input_dir, parallelism, chunk_size, chunk_overlap)

    # ── Step 2: collect to driver 
    log.info("=== Step 2/4: Collecting chunks to driver ===")
    all_chunks = chunk_rdd.collect()
    n_pdfs     = len({c["source_pdf"] for c in all_chunks})
    log.info("Collected %d chunks from %d PDF(s)", len(all_chunks), n_pdfs)

    # ── Step 3: embed 
    log.info("=== Step 3/4: Embedding (strategy=%s) ===", embedding_strategy)
    embed_fn   = EMBEDDING_DISPATCH[embedding_strategy]
    all_chunks = embed_fn(all_chunks, output_dir=output_dir, **embedding_kwargs)

    # ── Step 4: persist
    log.info("=== Step 4/4: Persisting to Parquet ===")
    save_to_parquet(spark, all_chunks, output_dir, embedding_strategy)

    elapsed = time.time() - t_start
    log.info("=== Pipeline complete in %.1fs ===", elapsed)

    return all_chunks
```

```python
# Adjust EMBEDDING_STRATEGY in the Configuration cell (cell 3) to switch
# between 'tfidf', 'word2vec', and 'sbert'.

chunks_out = run_pipeline(
    parallelism        = PARALLELISM,
    chunk_size         = CHUNK_SIZE,
    chunk_overlap      = CHUNK_OVERLAP,
    embedding_strategy = EMBEDDING_STRATEGY,
    input_dir          = INPUT_DIR,
    output_dir         = OUTPUT_DIR,
)

# Final summary 
print(f"\n{'='*60}")
print(f"  Pipeline summary")
print(f"{'='*60}")
print(f"  Total chunks produced : {len(chunks_out)}")
print(f"  Embedding strategy    : {chunks_out[0]['embedding_strategy']}")
print(f"  Embedding dimension   : {chunks_out[0]['embedding_dim']}")
print(f"  Output directory      : {OUTPUT_DIR}/strategy={EMBEDDING_STRATEGY}")

# Word count distribution
_wcs = [c["chunk_word_count"] for c in chunks_out]
print(f"  Chunk word count      : min={min(_wcs)}  mean={sum(_wcs)/len(_wcs):.0f}  max={max(_wcs)}")

# Unique PDFs
_pdfs = sorted({c["source_pdf"] for c in chunks_out})
print(f"  Source PDFs ({len(_pdfs)})      : {_pdfs}")
print(f"{'='*60}")
```
